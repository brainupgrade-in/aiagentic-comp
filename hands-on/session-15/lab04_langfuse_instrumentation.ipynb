## Step 1: LangFuse Trace Hierarchy

LangFuse organizes observability data in a three-level hierarchy:

```
Trace (top level -- one per user request)
  |
  +-- Span (logical step -- e.g., 'retrieve_context')
  |     |
  |     +-- Generation (LLM call -- input/output/tokens/cost)
  |
  +-- Span ('generate_answer')
        |
        +-- Generation (another LLM call)
```

| Level | What It Captures |
|---|---|
| Trace | User ID, session ID, total latency, tags |
| Span | Step name, start/end time, metadata |
| Generation | Model, prompt, completion, tokens, cost |

## Step 2: LangFuse Callback Handler Setup (LangChain)

LangFuse integrates with LangChain via a callback handler:

```python
from langfuse.callback import CallbackHandler

handler = CallbackHandler(
    public_key='pk-...',
    secret_key='sk-...',
    host='http://localhost:3000',    # LangFuse server
    trace_name='support-agent',
    user_id='user-123',
    session_id='session-abc',
)

# Pass to any LangChain invoke call:
result = chain.invoke(input, config={'callbacks': [handler]})
```

## Step 3: LangFuse Cost Tracking

LangFuse automatically tracks costs per generation:

| Field | Description |
|---|---|
| input_tokens | Number of prompt tokens |
| output_tokens | Number of completion tokens |
| total_tokens | input_tokens + output_tokens |
| input_cost | Cost of prompt (price * input_tokens) |
| output_cost | Cost of completion |
| total_cost | input_cost + output_cost |
| model | Model name (e.g., llama-3.1-70b) |

Cost formula (per generation):
```
input_cost  = input_tokens  * price_per_1k_input  / 1000
output_cost = output_tokens * price_per_1k_output / 1000
total_cost  = input_cost + output_cost
```
