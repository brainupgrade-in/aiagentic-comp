## Step 1: Python Health Checker Pattern

A Python HealthChecker monitors service health via HTTP probes:

| Field | Purpose |
|-------|--------|
| `url` | HTTP endpoint to probe (e.g., `/healthz`). HTTP 200 = healthy, other = unhealthy. |
| `interval_seconds` | Time between health checks (e.g., 30). |
| `timeout_seconds` | Max time to wait for response (e.g., 10). |
| `retries` | Consecutive failures before unhealthy. |
| `start_delay_seconds` | Grace period before first check. |

**Example HealthChecker config:**

```python
health_config = {
    "url": "http://localhost:8000/healthz",
    "interval_seconds": 30,
    "timeout_seconds": 10,
    "retries": 3,
    "start_delay_seconds": 15,
}
```

## Step 2: LangFuse Trace Monitoring

LangFuse provides LLM observability via its CallbackHandler.
It captures traces, spans, and generations automatically.

**Trace hierarchy:** `trace -> span -> generation`

| LangFuse Feature | What It Monitors |
|------------------|------------------|
| Traces | End-to-end request lifecycle |
| Spans | Individual steps (tool calls) |
| Generations | LLM calls with token counts |
| Cost tracking | Per-request and aggregate cost |
| Latency metrics | Response time per trace/span |
| Feedback scores | User satisfaction ratings |
