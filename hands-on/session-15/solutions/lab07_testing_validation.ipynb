## Step 1: Testing Pyramid for AI Agent Applications

Testing pyramid (bottom = most tests, top = fewest):

```
         /\
        /  \       E2E Tests (smoke tests, user flows)
       /----\
      /      \     Integration Tests (API + agent + DB)
     /--------\
    /          \   Unit Tests (functions, models, tools)
   /____________\
```

| Test Type | Count | What It Tests |
|-----------|-------|---------------|
| Unit | Many | Individual functions, pure logic |
| Integration | Some | Component interactions, API calls |
| E2E / Smoke | Few | Full user flow, critical paths |
| Load / Perf | Few | Throughput, latency under stress |

## Step 2: Service Level Agreements (SLAs)

Production AI agents need clear SLAs:

| Metric | Target | Measurement |
|--------|--------|-------------|
| Availability | 99.9% | Uptime over 30 days |
| Response time (p50) | < 2s | 50th percentile |
| Response time (p99) | < 10s | 99th percentile |
| Error rate | < 1% | 5xx responses / total |
| Throughput | 100 rps | Sustained requests/sec |
