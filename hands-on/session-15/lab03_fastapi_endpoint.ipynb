## Step 1: FastAPI Application Structure for AI Agents

A production FastAPI app for an AI agent typically has:

| Endpoint | Purpose |
|---|---|
| GET  /healthz | Liveness probe (health checker) |
| GET  /readyz | Readiness probe (startup sequencing) |
| POST /api/support | Submit a support question to agent |
| GET  /api/tickets | List support tickets |
| GET  /api/tickets/{id} | Get a specific ticket |
| GET  /metrics | Application metrics |

Response models use Pydantic BaseModel for validation:

```python
class SupportRequest(BaseModel):
    question: str
    customer_email: str | None = None
    session_id: str | None = None

class SupportResponse(BaseModel):
    answer: str
    ticket_id: str | None = None
    confidence: float
    sources: list[str]
```

## Step 2: Error Handling Patterns in FastAPI

Production error handling uses HTTPException with proper codes:

| Code | When | Example |
|---|---|---|
| 400 | Invalid request body | Missing required field |
| 404 | Resource not found | Ticket ID doesn't exist |
| 422 | Validation error | Wrong field type |
| 429 | Rate limit exceeded | Too many LLM calls |
| 500 | Internal server error | LLM API failure |
| 503 | Service unavailable | ChromaDB down |
