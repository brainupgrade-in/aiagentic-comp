## Step 3: Tools vs Resources — The Key Distinction

The difference is WHO INITIATES the interaction:

| Primitive | Controlled By | Description |
|---|---|---|
| Tool | The MODEL decides when to call it | Actions: query, create, send, update. May have side effects. |
| Resource | The APP decides when to load it | Context: schema, pages, config. Read-only, no side effects. |

**Example — Postgres MCP server:**
- Tool: `query_db(sql)` — model decides to run a query
- Resource: `db://schema` — app pre-loads table structure as context

**Example — Jira MCP server:**
- Tool: `create_issue()` — model decides to create a ticket
- Resource: `jira://board` — app pre-loads sprint board as context

The same data CAN be exposed as either — it depends on whether
the model or the application should control when it's accessed.
