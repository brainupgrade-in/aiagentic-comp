## Step 1: MCP Client Configuration Format

MCP clients (Claude Desktop, Claude Code) use JSON config files:

```json
{
  "mcpServers": {
    "server-name": {
      "command": "npx",
      "args": ["-y", "package-name"],
      "env": {"KEY": "value"}
    },
    "remote-server": {
      "url": "https://host/sse"
    }
  }
}
```

Two transport types:
- **stdio:** command + args (local subprocess)
- **sse:** url (remote HTTP)
