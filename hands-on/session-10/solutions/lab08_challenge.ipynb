## TODO 1 Solution: Python OTel Pipeline Setup

Create a Python script that configures a complete OTel tracing pipeline:
- Import: `trace`, `TracerProvider`, `BatchSpanProcessor`, `ConsoleSpanExporter`, `Resource`
- Create Resource with `service.name='agent-api'`, `service.version='2.0.0'`, `deployment.environment='production'`
- Create `ConsoleSpanExporter` instance
- Create `BatchSpanProcessor` wrapping the exporter
- Create `TracerProvider` with the resource
- Add the processor to the provider via `add_span_processor()`
- Set as global with `trace.set_tracer_provider()`
- Get a tracer with `trace.get_tracer('agent.api')`

````python
todo1_code = textwrap.dedent("""\
    from opentelemetry import trace
    from opentelemetry.sdk.trace import TracerProvider
    from opentelemetry.sdk.trace.export import (
        BatchSpanProcessor,
        ConsoleSpanExporter,
    )
    from opentelemetry.sdk.resources import Resource

    # Create Resource to identify the service
    resource = Resource.create({
        "service.name": "agent-api",
        "service.version": "2.0.0",
        "deployment.environment": "production",
    })

    # Create exporter (ConsoleSpanExporter for development)
    exporter = ConsoleSpanExporter()

    # Create processor (BatchSpanProcessor for efficiency)
    processor = BatchSpanProcessor(exporter)

    # Create and configure TracerProvider
    provider = TracerProvider(resource=resource)
    provider.add_span_processor(processor)
    trace.set_tracer_provider(provider)

    # Get a tracer
    tracer = trace.get_tracer("agent.api")
""")

with open(os.path.join(WORKDIR, "otel_setup.py"), "w") as f:
    f.write(todo1_code)
````

## TODO 2 Solution: Agent Launch Script with OTel + Structured Logging

Create a Python launch script for agent-api with OTel env vars and structured logging (no Docker required):
- `import os, logging, uvicorn`
- Set `OTEL_SERVICE_NAME`: `agent-api`
- Set `OTEL_EXPORTER_OTLP_ENDPOINT`: `http://localhost:4317`
- Set `OTEL_RESOURCE_ATTRIBUTES`: `deployment.environment=production,service.version=2.0`
- Set `OTEL_TRACES_SAMPLER`: `parentbased_traceidratio`
- Set `OTEL_TRACES_SAMPLER_ARG`: `0.1`
- Configure structured logging with JSON-like format
- Launch uvicorn on host `0.0.0.0`, port `8000`, log_level `info`

````python
todo2_code = textwrap.dedent("""\
    import os
    import logging
    import uvicorn

    # OTel environment configuration
    os.environ["OTEL_SERVICE_NAME"] = "agent-api"
    os.environ["OTEL_EXPORTER_OTLP_ENDPOINT"] = "http://localhost:4317"
    os.environ["OTEL_RESOURCE_ATTRIBUTES"] = "deployment.environment=production,service.version=2.0"
    os.environ["OTEL_TRACES_SAMPLER"] = "parentbased_traceidratio"
    os.environ["OTEL_TRACES_SAMPLER_ARG"] = "0.1"

    # Structured logging configuration
    logging.basicConfig(
        level=logging.INFO,
        format='{"time":"%(asctime)s","level":"%(levelname)s","logger":"%(name)s","message":"%(message)s"}',
    )
    logger = logging.getLogger("agent-api")
    logger.info("Starting agent-api with OTel environment configured")

    # Launch uvicorn
    if __name__ == "__main__":
        uvicorn.run(
            "app:app",
            host="0.0.0.0",
            port=8000,
            log_level="info",
        )
""")

with open(os.path.join(WORKDIR, "launch_agent.py"), "w") as f:
    f.write(todo2_code)
````

## Key Takeaways

This challenge covered the three pillars of a complete observability stack:

- **OTel Pipeline:** `TracerProvider` + `BatchSpanProcessor` + `ConsoleSpanExporter`
- **Launch Script:** `OTEL_*` env vars + structured logging + `uvicorn`
- **Design:** Metrics/logs/traces for all 4 AI components

Key patterns:
- Resource attributes identify your service (`service.name`, `service.version`, `deployment.environment`)
- `BatchSpanProcessor` batches spans for efficient export
- `ConsoleSpanExporter` is great for development; swap to OTLP exporter for production
- Structured logging (JSON format) makes logs machine-parseable
- Sampling (`parentbased_traceidratio` at 0.1) reduces overhead in production
- Each AI component (API, LLM, RAG, Tools) needs its own metrics, logs, and trace spans
