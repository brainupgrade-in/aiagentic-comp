# Lab 02: Message Types — Solution

**Goal:** Understand how SystemMessage and HumanMessage control LLM behavior.

```python
from langchain_ollama import ChatOllama
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

llm = ChatOllama(model="llama3.2:1b")
```

## Step 1: Without SystemMessage

```python
response = llm.invoke("Explain what an API is")
print(response.content)
```

## Step 2: With SystemMessage — kindergarten teacher

```python
messages = [
    SystemMessage(content="You are a kindergarten teacher. Explain everything using simple words and fun examples. Keep answers to 2 sentences."),
    HumanMessage(content="Explain what an API is"),
]
response = llm.invoke(messages)
print(response.content)
```

## Step 3: With SystemMessage — senior engineer

```python
messages = [
    SystemMessage(content="You are a senior software engineer. Use technical terms. Be precise and concise. One sentence only."),
    HumanMessage(content="Explain what an API is"),
]
response = llm.invoke(messages)
print(response.content)
```

## TODO 3: Build a Message Inspector

```python
from collections import Counter

sample_messages = [
    SystemMessage(content="You are a helpful assistant. Be concise."),
    HumanMessage(content="What is Python?"),
    AIMessage(content="Python is a high-level, interpreted programming language known for its simplicity."),
    HumanMessage(content="What about Java?"),
    AIMessage(content="Java is a compiled, object-oriented language popular in enterprise applications."),
]


def inspect_messages(messages):
    """Analyze a list of LangChain messages."""
    type_counts = Counter()
    type_lengths = Counter()
    type_roles = {}

    for msg in messages:
        name = type(msg).__name__
        type_counts[name] += 1
        type_lengths[name] += len(msg.content)
        type_roles[name] = msg.type  # .type gives the role string

    print(f"Total messages: {len(messages)}\n")
    for msg_type in type_counts:
        print(f"  {msg_type}:")
        print(f"    count:          {type_counts[msg_type]}")
        print(f"    total chars:    {type_lengths[msg_type]}")
        print(f"    role (.type):   '{type_roles[msg_type]}'")


inspect_messages(sample_messages)
```

## Key Takeaways

- SystemMessage controls HOW the LLM responds
- Same question + different SystemMessage = different answer
- You can enforce persona, tone, format, and rules
- Message lists simulate conversation history
