# Lab 01: Hello LangChain — Solution

**Goal:** Connect to a local LLM (Ollama) using LangChain and send your first message.

**What you'll learn:**
- How to import and create a ChatOllama model
- How to use `.invoke()` to send a message
- What the response object looks like (AIMessage)

## Step 1: Create a connection to the local Ollama LLM

ChatOllama connects to the Ollama server running on your machine.
We use `llama3.2:1b` — a small but capable model.

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(model="llama3.2:1b")

print("Connected to Ollama!")
```

## Step 2: Send a simple message using `.invoke()`

`.invoke()` sends a message and waits for the complete response.

```python
response = llm.invoke("What is Python in one sentence?")

print("--- Full Response Object ---")
print(f"Type: {type(response)}")
print(f"Content: {response.content}")
print(f"Metadata: {response.response_metadata}")
```

## Step 3: The response is an AIMessage object

The key property is `.content` — that's the text the LLM generated.

```python
print("--- Just the text ---")
print(response.content)
```

## TODO 1: Explore the AIMessage Object

```python
response = llm.invoke("What is Kubernetes in one sentence?")

print(f"content:           {response.content[:80]}...")
print(f"type:              {response.type}")
print(f"id:                {response.id}")
print(f"response_metadata: {response.response_metadata}")
print(f"  model:           {response.response_metadata.get('model', 'N/A')}")
print(f"  eval_count:      {response.response_metadata.get('eval_count', 'N/A')} tokens generated")
print(f"  eval_duration:   {response.response_metadata.get('eval_duration', 0) / 1e9:.2f}s")

# What else is inside?
print(f"\nAll attributes: {[a for a in dir(response) if not a.startswith('_')]}")
```

## TODO 2: Compare Model Parameters

```python
llm_precise = ChatOllama(model="llama3.2:1b", temperature=0)
llm_creative = ChatOllama(model="llama3.2:1b", temperature=1.5)

question = "Suggest a name for a Python logging library"

print("--- temperature=0 (precise, deterministic) ---")
for i in range(2):
    r = llm_precise.invoke(question)
    print(f"  Run {i+1}: {r.content}")

print("\n--- temperature=1.5 (creative, varied) ---")
for i in range(2):
    r = llm_creative.invoke(question)
    print(f"  Run {i+1}: {r.content}")

# Also try num_predict to limit output length
llm_short = ChatOllama(model="llama3.2:1b", num_predict=15)
r = llm_short.invoke("Explain what Docker is")
print(f"\n--- num_predict=15 (truncated) ---\n  {r.content}")
```

## Key Takeaways

- `ChatOllama` connects LangChain to your local Ollama
- `.invoke()` sends a message and returns an `AIMessage`
- `AIMessage` has `.content`, `.response_metadata`, `.id`, and more
- Model parameters like `temperature` and `num_predict` control output behavior
