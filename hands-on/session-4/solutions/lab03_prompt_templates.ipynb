# Lab 03: Prompt Templates -- Solution

**Goal:** Build reusable prompts with variables using ChatPromptTemplate.

````python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate

llm = ChatOllama(model="llama3.2:1b")
````

## Step 1: Translation Template

````python
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a translator. Translate accurately and only return the translation."),
    ("human", "Translate '{text}' to {language}"),
])

for text, lang in [("Good morning", "French"), ("Good morning", "Japanese"), ("Thank you", "Hindi")]:
    messages = prompt.invoke({"text": text, "language": lang})
    response = llm.invoke(messages)
    print(f"  '{text}' -> {lang}: {response.content}")
````

## Step 2: Look inside the template

````python
messages = prompt.invoke({"text": "Hello", "language": "Spanish"})
print(f"Type: {type(messages)}")
print(f"Messages: {messages.to_messages()}")
````

## Step 3: Multi-variable template

````python
explain_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a {role}. Explain concepts at a {level} level. Keep it to {length}."),
    ("human", "Explain {topic}"),
])

response = llm.invoke(
    explain_prompt.invoke({
        "role": "friendly teacher",
        "level": "beginner",
        "length": "2 sentences",
        "topic": "what a variable is in programming",
    })
)
print(response.content)
````

## TODO 1: Code review template

````python
review_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a senior {language} code reviewer. Review the code for correctness, style, and potential issues. Be concise."),
    ("human", "Review this code:\n```\n{code}\n```"),
])

response = llm.invoke(review_prompt.invoke({
    "language": "Python",
    "code": "def add(a, b): return a + b",
}))
print(response.content)
````

## TODO 2: Email writer template

````python
email_prompt = ChatPromptTemplate.from_messages([
    ("system", "You write short, professional emails. Tone: {tone}. Keep it under 5 sentences."),
    ("human", "Write an email to {recipient} about: {subject}"),
])

response = llm.invoke(email_prompt.invoke({
    "tone": "formal",
    "recipient": "the engineering team",
    "subject": "scheduled maintenance this weekend",
}))
print(response.content)
````

## TODO 3: Partial Variables — Pre-fill Template Values

````python
# explain_prompt has variables: {role}, {level}, {length}, {topic}
# Create a "beginner teacher" prompt that only needs {topic}
beginner_prompt = explain_prompt.partial(
    role="friendly teacher",
    level="beginner",
    length="2 sentences",
)

response = llm.invoke(beginner_prompt.invoke({"topic": "what a container is"}))
print(f"Beginner: {response.content}\n")

# Create an "expert" version
expert_prompt = explain_prompt.partial(
    role="senior architect",
    level="expert",
    length="1 sentence",
)
response = llm.invoke(expert_prompt.invoke({"topic": "what a container is"}))
print(f"Expert: {response.content}")

# Check: how many variables does each prompt need?
print(f"\nOriginal variables:  {explain_prompt.input_variables}")
print(f"Beginner variables:  {beginner_prompt.input_variables}")
print(f"Expert variables:    {expert_prompt.input_variables}")
````

## Lab 03 complete!

**Key takeaways:**
- `ChatPromptTemplate` makes prompts reusable with `{variables}`
- `.invoke()` fills in the variables and returns formatted messages
- Templates enforce consistency across different inputs
- `from_messages()` for system+human, `from_template()` for simple prompts
- `.partial()` pre-fills variables to create specialized template variants
