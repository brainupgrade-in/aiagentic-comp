# Lab 08 Challenge: Technical Knowledge Assistant -- Solution

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = ChatOllama(model="llama3.2:1b")
```

## Part A: Explanation Chain

```python
explain_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a technical educator. Explain concepts clearly and concisely for a professional audience. Keep it to 2-3 sentences."),
    ("human", "Explain {technology} in a clear, professional summary."),
])
explain_chain = explain_prompt | llm | StrOutputParser()
```

## Part B: Question Generator Chain

```python
question_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a technical educator. Generate one thought-provoking knowledge-check question to test understanding. Return just the question, nothing else."),
    ("human", "Generate a knowledge-check question about {technology}"),
])
question_chain = question_prompt | llm | StrOutputParser()
```

## Part C: Answer Chain

```python
answer_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a senior engineer providing a detailed reference answer. Give a clear, structured answer in 3-4 sentences."),
    ("human", "Answer this question:\n{question}"),
])
answer_chain = answer_prompt | llm | StrOutputParser()
```

## Part D: Run the Pipeline

```python
technologies = ["Flask", "FastAPI", "REST API"]

for technology in technologies:
    print(f"\nTechnology: {technology}")

    explanation = explain_chain.invoke({"technology": technology})
    print(f"\nExplanation:\n{explanation}")

    question = question_chain.invoke({"technology": technology})
    print(f"\nKnowledge Check:\n{question}")

    answer = answer_chain.invoke({"question": question})
    print(f"\nReference Answer:\n{answer}")
    print()
```
