**What you'll learn:**
- How to connect to Groq's cloud LLM using ChatGroq

- That LangChain's model-agnostic design means only the import changes

```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain_core.messages import SystemMessage, HumanMessage

load_dotenv()
```

## Step 1: Connect to Groq

```python
llm = ChatGroq(model="llama-3.3-70b-versatile")
print("Connected to Groq!")
```

## Step 2: Send your first message via Groq

```python
response = llm.invoke("What is RAG in AI? Answer in 2 sentences.")

print("--- Response from Groq (70B model) ---")
print(f"Content: {response.content}")
print(f"\nModel used: {response.response_metadata.get('model_name', 'unknown')}")
```

## Step 3: Test with structured messages

```python
response = llm.invoke([
    SystemMessage(content="You are a helpful AI assistant. Be concise."),
    HumanMessage(content="Explain embeddings in AI in one sentence."),
])

print("--- Structured message response ---")
print(response.content)
```

## Step 4: Notice the quality difference

```python
response = llm.invoke([
    SystemMessage(content="You are a Python expert. Be concise but accurate."),
    HumanMessage(content="What are the top 3 differences between a list and a tuple in Python?"),
])

print("--- Quality test ---")
print(response.content)
```

## TODO 1: Explore Groq's Response Metadata

```python
response = llm.invoke("Explain what RAG is in 3 sentences.")

meta = response.response_metadata
usage = meta.get("token_usage", {})

print(f"--- Groq Response Metadata ---")
print(f"Model:              {meta.get('model_name', 'N/A')}")
print(f"Prompt tokens:      {usage.get('prompt_tokens', 'N/A')}")
print(f"Completion tokens:  {usage.get('completion_tokens', 'N/A')}")
print(f"Total tokens:       {usage.get('total_tokens', 'N/A')}")
print(f"Queue time:         {usage.get('queue_time', 'N/A')}s")
print(f"Prompt time:        {usage.get('prompt_time', 'N/A')}s")
print(f"Completion time:    {usage.get('completion_time', 'N/A')}s")
print(f"Total time:         {usage.get('total_time', 'N/A')}s")

# Calculate tokens per second
comp_time = usage.get("completion_time", 0)
comp_tokens = usage.get("completion_tokens", 0)
if comp_time and comp_time > 0:
    print(f"\nSpeed: {comp_tokens / comp_time:.0f} tokens/sec (Groq is FAST!)")
```

## TODO 2: Test with a different Groq model (SOLVED)

Compare the 8B instant model with the 70B versatile model.

```python
llm_fast = ChatGroq(model="llama-3.1-8b-instant")
fast_response = llm_fast.invoke("Explain Python virtual environments in 2 sentences.")
print(f"\n8B model: {fast_response.content}")

smart_response = llm.invoke("Explain Python virtual environments in 2 sentences.")
print(f"\n70B model: {smart_response.content}")
```

## Key Takeaways

- ChatGroq connects to Groq's cloud API (free, fast)
- Same `.invoke()` API as ChatOllama — only import changes
- 70B model is significantly better than 1B for complex tasks
- `GROQ_API_KEY` must be set as an environment variable
