```python
print("LangFuse organizes data in three levels:\n")
print("  Trace  \u2500\u2500\u2500 top-level, one per user request")
print("  \u251c\u2500\u2500 Span  \u2500\u2500\u2500 logical grouping (rag_retrieval, tool_execution)")
print("  \u2502   \u251c\u2500\u2500 Generation  \u2500\u2500\u2500 actual LLM API call")
print("  \u2502   \u2514\u2500\u2500 Generation  \u2500\u2500\u2500 another LLM call")
print("  \u2514\u2500\u2500 Span  \u2500\u2500\u2500 another step")
print("      \u2514\u2500\u2500 Generation  \u2500\u2500\u2500 LLM call")
```

## Key Takeaways

- **LangFuse fills the AI observability gap** that traditional tools (OTel, Prometheus) cannot cover: exact prompt text, full LLM responses, cost tracking, and user feedback
- **Trace hierarchy** organizes LLM observability data in three levels: Trace (per request) > Span (logical grouping) > Generation (actual LLM API call)
- **Complementary stack:** OTel handles infrastructure metrics and distributed tracing, while LangFuse handles AI-specific concerns like prompt management and cost tracking
- **Self-hosted option** allows running LangFuse on your own infrastructure with a PostgreSQL backend for full data control
- **Tool selection matters:** Use `otel_metrics` for system health, `opentelemetry` for distributed tracing, and `langfuse` for LLM-specific observability
